# Otopilot

```python
def heatCheck(dataFrame, plot):
    if plot.__name__ == "heatmap":
        plt.figure(figsize=(10, 8))
        plot(data=dataFrame.select_dtypes("number").corr(), annot=True)
        plt.xticks(rotation=90)
        return True

def pairCheck(dataFrame, plot, target):
    if plot.__name__ == "pairplot":
        plt.figure(figsize=(15, 10))
        plot(data=dataFrame, hue=target, corner=True, diag_kind="kde")
        plt.xticks(rotation=90)
        return True

def barCheck(dataFrame, plot, target, column):
    if plot.__name__ == "barplot":
        plt.figure(figsize=(6, 4))
        ax = plot(data=dataFrame, x=target, y=column, palette="viridis")
        ax.bar_label(ax.containers[0])  # Barların üzerine değerleri yazdırma
        plt.title(target + " & " + column + " " + plot.__name__)
        plt.show()
        return True

def plotCheck(dataFrame, plot, target):
    return pairCheck(dataFrame, plot, target) or heatCheck(dataFrame, plot)

def otoPlot(dataFrame, plots=[sns.barplot, sns.scatterplot, sns.boxplot, sns.pairplot, sns.heatmap], target=None, include="number"):
    columns = dataFrame.select_dtypes(include).columns if include != None else dataFrame.columns
    hedef, columns = ("", columns) if target == None else (target + " & ", columns.drop(target))
    for plot in plots:
        if plotCheck(dataFrame, plot, target):
            continue
        for column in columns:
            if barCheck(dataFrame, plot, target, column):
                continue
            plot(data=dataFrame, x=target, y=column)
            plt.title(hedef + column + " " + plot.__name__)
            plt.show()

# Kullanım örneği: otoPlot(veriSeti, target='hedefSütunu')
```
